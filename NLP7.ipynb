```python
import nltk
from nltk import pos_tag,ne_chunk, word_tokenize
from nltk.chunk import RegexpParser
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('tagsets_json')
from nltk.tree import Tree
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
from nltk.chunk import RegexpParser
```

```python
sentence = "The smart student solved the difficult problem."
sentence1 = "This is the class of natural language processing."
```

```python
tokens = word_tokenize(sentence1)
tagged = pos_tag(tokens)
tagged
```

```python
nltk.help.upenn_tagset("NNP")
```

```python

named_entities = ne_chunk(tagged)
print (named_entities)
```

```python
for subtree in named_entities.subtrees():
    if isinstance(subtree, nltk.Tree):

        entity = "   ".join([word for word, tag in subtree.leaves()])
        label = subtree.label()
        print(f"Entity: {entity}, Label: {label}")
```

```python
grammar = "NP: {<DT>?<JJ>*<NN>}"
```

```python
chunk_parser = RegexpParser(grammar)
chunked = chunk_parser.parse(tagged)
print(chunked)
```

```python
grammar1 = "VP: {<MD>?<VB.*><RB.*>*<VB.*>*}"

# Initialize the chunk parser with the rule
chunk_parser = RegexpParser(grammar1)

# Parse your POS-tagged text (assuming 'tagged' is defined)
chunked1 = chunk_parser.parse(tagged)

# Display the text representation
print(chunked1)

# Open a GUI window to visualize the parse tree
chunked1.draw()
```
